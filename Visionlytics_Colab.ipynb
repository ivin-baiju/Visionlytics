```python
# 3. Run Streamlit in the background and expose it via Cloudflare Tunnel
import subprocess
import time

# Start Streamlit
subprocess.Popen(['streamlit', 'run', 'app/main.py'])

# Give Streamlit a few seconds to boot up
time.sleep(5)

# Start the Cloudflare Tunnel
print("\n\n====== YOUR APP URL IS GENERATING BELOW ======\n")
!./cloudflared-linux-amd64 tunnel --url http://localhost:8501
```
